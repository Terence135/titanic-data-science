```python
df2['familySize'] = df2['SibSp'] + df2['Parch'] + 1
```

```python
(df2.groupby('familySize').agg(passCount=('PassengerId', 'size'),
                             survival=('Survived', 'mean'))
    .assign(survival=lambda x: x['survival'] * 100))
```

```python
df2['familyType'] = pd.cut(df2['familySize'], bins= [1, 2, 5, float('inf')], labels= ['Alone', 'Small', 'Large'],
                          right=False )
```

```python
df2.groupby('familyType').agg(PasssC=('PassengerId', 'size'), 
                             survivPer=('Survived', 'mean')).assign(survivPer= lambda x: x['survivPer'] * 100)
```

```python
df2['AgeBrand'] = pd.cut(df2['Age'], bins=[0,18,40,60,float('inf')], labels=['0-18','18-39','40-59','60+'],
                         right=False)
```

```python
df2.groupby(['AgeBrand', 'Pclass']).agg(survivalPer=('Survived', 'mean')
                                       ).assign(survivalPer=lambda x: x['survivalPer'] * 100).unstack()
```
